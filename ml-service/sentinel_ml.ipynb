```python
import pandas as pd

df = pd.read_csv('train.csv')  # path to where you saved it
print(df['label'].value_counts())
print(df.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df['text'], df['label'],
    test_size=0.2,
    random_state=42
)

print(f"Train size: {len(X_train)}, Test size: {len(X_test)}")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

model = Pipeline([
    ('tfidf', TfidfVectorizer(max_features=5000, ngram_range=(1,2))),
    ('clf', LogisticRegression(max_iter=1000))
])

model.fit(X_train, y_train)
print("Training done!")
```

```python
from sklearn.metrics import accuracy_score, classification_report

y_pred = model.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2%}")
print(classification_report(y_test, y_pred, target_names=['Human', 'AI']))
```

```python
import joblib

joblib.dump(model, 'sentinel_model.pkl')
print("Model saved!")
```

```python
tests = [
    "i honestly had no clue what the answer was so i just guessed lol",
    "The results demonstrate a statistically significant improvement across all evaluated metrics."
]

for t in tests:
    label = model.predict([t])[0]
    confidence = max(model.predict_proba([t])[0])
    print(f"→ {'AI' if label == 1 else 'Human'} ({confidence:.2%}) | {t[:60]}...")
```
